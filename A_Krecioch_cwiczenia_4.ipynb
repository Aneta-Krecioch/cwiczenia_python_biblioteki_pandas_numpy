#### KROK 1 – Wczytanie i wstępna analiza danych

```python
#definicja df
df = pd.DataFrame({
    "produkt": ["A", "a", "B", None, "C", "C"],
    "produkcja": [120, 0, 95, 110, -5, 130],
    "plan": [100, 100, None, 105, 120, 125],
    "koszt": ["5000", "4800", None, "5200", "99999", "5100"]
})

print(df)
print(df.shape)
print(df.dtypes)
print(df.head())
```

#### KROK 2 – Identyfikacja problemów w danych 

znajdź potencjalne błędy:
<ul>
<li>wartości ujemne w produkcji
<li>wartości równe 0
<li>nietypowe wartości (np. bardzo wysokie koszty)
</ul>

```python
df.isnull().sum()
```

```python
print(df["produkt"].unique)
```

#### KROK 3 – Czyszczenie danych

produkt, plan, koszt ma null

```python
df["koszt"] = df["koszt"].astype(float)
df['koszt'] = df['koszt'].replace(0, np.nan)
srednia_koszt = df['koszt'].mean()
df['koszt'] = df['koszt'].fillna(srednia_koszt)
df['plan'] = df['plan'].replace(0, np. nan)
srednia_plan = df['plan'].mean()
df['plan'] = df['plan'].fillna(srednia_plan)
df["produkt"] = df["produkt"].str.upper()
df = df.query('produkcja > 0')
df = df.dropna()
print(df)
```

#### KROK 4 – Tworzenie wskaźników KPI

```python
df["wydajnosc"] = df["produkcja"] / df["plan"] * 100
df["koszt_jednostkowy"] = df["koszt"] / df["produkcja"]
print(df)
```

#### KROK 5 – Analiza danych

```python
# statystyki
df.describe()
print(f"Średnia proukcji: {df.groupby("produkt")["produkcja"].mean()}")
print(f"Suma kosztów: {df.groupby("produkt")["koszt"].sum()}")
```

#### KROK 6 – Identyfikacja anomalii

```python
df_max = df["koszt"].max()
df_min = df["produkcja"].min()
print(f"Maksimum kosztu: {df_max}")
print(f"Minimum produkcja: {df_min}")
print(df)
```

```python
df_filtr = df[df["koszt"] > 10000]
df_mediana = df_filtr["koszt"].median()
df_średnia = df_filtr["koszt"].mean()
df_max = df_filtr["koszt"].max()
df_min = df_filtr["koszt"].min()
print(f"Mediana: {df_mediana}")
print(f"Średnia: {df_średnia}")
print(f"Max: {df_max}")
print(f"Min: {df_min}")
```

## Część 2: biblioteka pandas

```python
#utworzenie danych tabelarycznych
df = pd.DataFrame({
    "produkt": ["A", "B", "A", "B", "C"],
    "produkcja": [100, 120, 90, 130, 80],
    "koszt": [5000, 6000, 4500, 6500, 4000]
})

#wyświetlenie danych
print("Wyświetlenie danych")
print(df)
#sprawdzenie typów danych
print("Typy danych")
print(df.dtypes)
#pierwszy wiersz danych
print("Pierwszy wiersz danych")
print(df.head(1))
#wyświetlenie kształtu tabeli
print("Kształt tabeli")
print(df.shape)
```

Tabela zawiera **5 wierszy** i **3 kolumny**. <br>
Jedna kolumna **string** (produkt) i dwie typu **integer** (produkcja i koszt)

```python
#tworzenie nowej kolumny w tabeli
df['sprzedaż'] = df['produkcja']*df['koszt']
print(df)
```

```python
#filtrowanie danych
df = df[df['produkcja'] > 100]
print(df)
```

```python
#sortowanie danych
df=df.sort_values(['produkcja','koszt'], ascending=[False, True])
print(df)
```

```python
#grupowanie danych
srednia_produkcja = df.groupby("produkt")["produkcja"].mean()
srednia_koszt = df.groupby("produkt")["koszt"].mean()
#wyświetlenie wyników
print(f"Dane o średniej produkcji {srednia_produkcja}")
print(f"Dane o średniej koszt {srednia_koszt}")
```

```python
#dodanie koszt jednostkowy
df['koszt jednostkowy'] = df['koszt']/df['produkcja']
print(df)
```
