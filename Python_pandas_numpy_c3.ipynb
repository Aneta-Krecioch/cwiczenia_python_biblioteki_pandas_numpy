```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'produkt': ['A', 'b', None, 'C', 'a'],
    'produkcja': [100, 0, 120, -5, 90],
    'plan': [120, 100, np.nan, 110, 100],
    'koszt': ['5000', '4500', '5200', None, '4800']
})

print(df)
```

#### Sprawdzanie braków danych

```python
#wykrywanie pustych wartości
df.isnull().sum()
```

```python
df_null = df.isnull().sum()
df_null_max = df_null.max()

df_kolumny = df_null[df_null == df_null_max].index.tolist()
print(f"Kolumna z największą liczbą nulli: {df_kolumny}")
```

#### Usuwanie brakujących danych

```python
#metoda dropna()
df_dropna = df.dropna()
print(df_dropna)
```

```python
rekordy_przed = df.shape[0]
rekordy_po = df_dropna.shape[0]

if rekordy_przed == rekordy_po:
    print('Liczba rekordów jest taka sama')
else:
    print('Liczba rekordów różna')
```

#### Uzupełnianie braków danych

```python
#fillna()
df_fill = df.fillna(0)
print(df_fill)
```

```python
#porównanie dropna() i fillna(0)
rekordy_fill = df_fill.shape[0]
if rekordy_przed == rekordy_po:
    print('Liczba rekordów po drop się nie zmieniła')
elif rekordy_przed == rekordy_fill:
    print('Liczba rekordów po drop się zmieniła ale po fill nie')
else:
    print('Liczba rekordów różna w obu metodach')
```

#### Tworzenie nowej kolumny

```python
df['plan'] = df['plan'].fillna(100)
df['wydajnosc'] = (df['produkcja']/df['plan']*100).round(2)
print(df)
```

#### Zmiana typu danych

```python
print(f"Typ kolumny koszt przed zmianą: {df['koszt'].dtype}")
df['koszt'] = df['koszt'].fillna('0')
df['koszt'] = df['koszt'].astype(float)
print(f"Typ kolumny koszt po zmianie: {df['koszt'].dtype}")
```

#### Usuwanie błędnych rekordów

```python
print(f"Tabela przed usunięciem")
print(df)

print(f"Tabela po usunięciu")
df = df[df['produkcja'] > 0]
print(df)
```

#### Standaryzacja danych tekstowych

```python
#uzupełnienie wartości null w kolumnie produkt
df['produkt'] = df['produkt'].fillna('BRAK')

#standaryzacja zapisu
df['produkt'] = df['produkt'].str.upper()

#wyświetlenie df
print(df)
```

#### Łączenie kilku operacji

```python
#pełny proces przygotowania danych
df = pd.DataFrame({
    'produkt': ['A', 'b', None, 'c', 'd', 'B'],
    'produkcja': [100, 0, -5, 90, 240, 90],
    'plan': [120, 0, np.nan, 100, 240, 50],
    'koszt': ['120', '0', None, '100', '240', '50'],
    'czas': ['','120','50','12',None,'123'],
    'magazyn': [140, 30, 190, np.nan, -10, 50],
})

#dane wstępne 
print(f"Dane początkowe: {df}")
```

```python
#sprawdzenie ilości braków w danych
print(f"Braki danych: {df.isnull().sum()}")

#output: jeśli w zbiorze danych jest pusty ciąg znaków funkcja isnull go nie wskazuje
#co oznacza, że funkcja fillna() go nie wypełni - trzeba zrobić to w inny sposób
#np. zmienić pusty ciąg znaków na null a później na całości użyć funkcji fillna()
```

```python
#zamiana pustych ciągów znaków na wartości null
df = df.replace(r'^\s*$', np.nan, regex=True)

#sprawdzenie ile null
print(f"Braki danych: {df.isnull().sum()}")
print(df.dtypes)
```

```python
#uzupełnienie braków w danych

#kolumny liczbowe
df['plan'] = df['plan'].fillna(0)
df['magazyn'] = df['magazyn'].fillna(0)

#kolumny tekstowe
df['produkt'] = df['produkt'].fillna('BRAK')
df['koszt'] = df['koszt'].fillna('BRAK')
df['czas'] = df['czas'].fillna('BRAK')

#wyświetlenie uzupełnionego df
print(df)
```

```python
#wykluczenie danych dla produkcji <=0
df = df[df['produkcja'] > 0]

#przypisanie poprawnych typów danych
df['koszt'] = df['koszt'].astype(float)

#ujednolicenie zapisu
df['produkt'] = df['produkt'].str.upper()

#dodanie brakującej kolumny wydajność
df['wydajnosc'] = df['produkcja'] / df['plan'].replace(0,
np.nan) * 100

#oczyszczone dane
print(f"Dane po przygotowaniu: {df}")
```
